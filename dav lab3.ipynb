```python
import pandas as pd

# Create your first dataframe
df = pd.DataFrame({
    "Order Date": ["2024-01-10", "2024-01-15"],
    "Sales": [50000, 2500]
})

# Create your second dataframe
new_sales = pd.DataFrame({
    "Order Date": ["2024-02-05", "2024-02-12"],
    "Sales": [9000, 30000]
})

# Combine them cleanly
combined_df = pd.concat([df, new_sales], ignore_index=True)

print("CONCATENATED DATA:")
display(combined_df)
```

```python
import pandas as pd

# 1. Define df
df = pd.DataFrame({
    "Order Date": ["2024-01-10", "2024-01-15", "2024-02-05"],
    "Product Name": ["Laptop", "Mouse", "Printer"],
    "Sales": [50000, 2500, 9000]
})

# 2. Define product_info with the exact same Product Name column
product_info = pd.DataFrame({
    "Product Name": ["Laptop", "Mouse", "Printer"],
    "Discount": [10, 5, 15]
})

# 3. Merge them
merged_df = pd.merge(df, product_info, on="Product Name", how="left")

print("MERGED DATA:")
display(merged_df)
```

```python
product_info = pd.DataFrame({ "Product Name": [ "Laptop", "Mouse", "Printer", "Tablet", "Monitor" ], "Supplier": [ "Dell", "Logitech", "HP", "Samsung", "LG" ], "Discount": [ 10, 5, 15, 8, 12 ] }) 
print("PRODUCT INFORMATION") 
display(product_info)
merged_df = pd.merge( df, product_info, on="Product Name", how="left" ) 
print("MERGED DATA") 
display(merged_df) 
```

```python
region_info = pd.DataFrame( {  "Region_Manager": [ "Manager A", "Manager B", "Manager C", "Manager D" ] },
                            index=[ "North", "South", "East", "West" ] ) 
print("REGION INFORMATION") 
display(region_info)
df_region = df.set_index("Region") 
joined_df = df_region.join(region_info)
print("JOINED DATA")
display(joined_df) 
```

```python
import pandas as pd

# Define the dataframe directly to avoid file path errors
df = pd.DataFrame({
    "Order Date": ["2024-01-10", "2024-01-15", "2024-02-05", "2024-02-12"],
    "Region": ["North", "South", "East", "West"],
    "Sales": [50000, 2500, 9000, 30000]
})

region_sales = df.groupby("Region")["Sales"].sum()
print("TOTAL SALES BY REGION")
display(region_sales)
```

```python
import pandas as pd

# Define df with all required columns so pandas finds them
df = pd.DataFrame({
    "Category": ["Electronics", "Accessories", "Office", "Electronics"],
    "Region": ["North", "South", "East", "West"],
    "Sales": [50000, 2500, 9000, 30000]
})

pivot_table = pd.pivot_table(df, values="Sales", index="Category", columns="Region", aggfunc="sum", fill_value=0)
print("PIVOT TABLE")
display(pivot_table)
```
